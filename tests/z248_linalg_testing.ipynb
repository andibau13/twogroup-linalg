## test if kernel vectors multiply to zero for filtration

```python
for i in range(100):
    A = lin.z248_hom.rand_dim(10)
    ker = A.kernel()
    print(f"{A.dim0} x {A.dim1}: {(A@ker).is_zero()}")
```

## Test for snf kernel method

```python
for i in range(100):
    A = lin.z248_hom.rand_dim(7)
    try:
        ker = snf.z248_z248_kernel(A)
        print(f"{A.dim0} x {A.dim1}: {(A@ker).is_zero()}")
    except:
        print(f"{A.dim0} x {A.dim1}: Snf kernel method overflow")
```

## Test if filtration and snf kernel have the same dimensions (form same groups)

```python
for i in range(100):
    A = lin.z248_hom.rand_dim(5)
    ker_filt = A.kernel()
    try:
        ker_snf = snf.z248_z248_kernel(A)
        print(f"{A.dim0} x {A.dim1}: {ker_filt.dim1 == ker_snf.dim1}")
    except:
        print(f"{A.dim0} x {A.dim1}: Snf kernel method overflow")
```

## Testing if z248_solve_helper does indeed yield a solution

```python
for i in range(100):
    X = lin.z248_hom.rand_dim(5)
    k_to_find = lin.z248_elem.rand(X.dim1)
    #b = lin.z248_elem.rand(X.dim0)
    b = X @ k_to_find
    ker, helper = X.kernel(return_solve_helper = True)
    try:
        k = lin.z248_solve_with_helper(X, b, helper)
        print(f"Xk=b:{(X @ k - b).is_zero()}")
    except:
        print("Solution not found!?")
```

## Testing epi-mono decomposition

```python
for i in range(1000):
    X = lin.z248_hom.rand_dim(5)
    L, R = lin.z248_epi_mono(X)
    Lker = L.kernel()
    Xker = X.kernel()
    X_coker_dim = X.dim1[0]+2*X.dim1[1]+3*X.dim1[2] - (Xker.dim1[0]+2*Xker.dim1[1]+3*Xker.dim1[2])
    X_img_dim = L.dim1[0]+2*L.dim1[1]+3*L.dim1[2]
    print(f"LR=X:{(L @ R - X).is_zero()}, L injective: {Lker.dim1 == (0,0,0)}, Cokernel dim = Image dim: {X_coker_dim == X_img_dim}")
```

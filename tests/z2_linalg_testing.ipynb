```python
import third_order_gates.z2_helpers as z2lin
import numpy as np
```

## Test solving Z2 linear equations with helpers

```python
for i in range(1000):
    shape = np.random.randint(0,10,size=(2,))
    A = np.random.randint(0,2,size=shape)
    sol_to_find = np.random.randint(0,2,size=(A.shape[1],))
    b = A @ sol_to_find
    help = z2lin.get_solve_helper(A)
    try:
        sol = z2lin.solve_with_helper(*help, b)
        print(np.all((A @ sol - b) % 2 == 0))
    except:
        print("Solution not found!")
```

## Testing bitgauss functions

```python
import third_order_gates.bitgauss_wrappers as bitg
import third_order_gates.galois_wrappers as galo
for i in range(20):
    dim0, dim1 = np.random.randint(0,5,size=(2,))
    mat = np.random.randint(0,2,size=(dim0, dim1))
    gauss_ker = bitg.kernel(mat)
    gauss_ker_test = np.all((mat @ gauss_ker) % 2 == 0)
    galois_ker = galo.kernel(mat)
    kernel_dims_same = gauss_ker.shape[0] == galois_ker.shape[0]
    print(f"bitgauss kernel test: {gauss_ker_test}, kernel dimension equal to galois kernel: {kernel_dims_same}")
```

```python
for i in range(20):
    dim0, dim1 = np.random.randint(0,5,size=(2,))
    mat = np.random.randint(0,2,size=(dim0, dim1))
    gauss_red = bitg.rref(mat)
    galois_red = galo.rref(mat)
    row_reduction_same = np.all(gauss_red - galois_red == 0)
    print(f"bitgauss and galois yield same row reduction: {row_reduction_same}")
```

## Comparing kernel to z248 spezialized kernel

```python
for i in range(1000):
    x = lin.Hom.rand_dim(4, 3, 3)
    xpr = z248lin.z248_hom(x.M, x.dim0, x.dim1)
    xker = x.kernel()
    xprker = xpr.kernel()
    xprkertest = lin.Hom(xprker.M, xprker.dim0, xprker.dim1)
    print((xker - xprkertest).is_zero())
```

## Testing kernel

```python
for i in range(100):
    x = lin.Hom.rand_dim(4, 5, 7)
    xker = x.kernel()
    print((x@xker).is_zero())
```

## Testing solve_with_helper

```python
for i in range(100):
    x = lin.Hom.rand_dim(3, 5, 5)
    sol_to_find = lin.Elem.rand(x.dim1)
    b = x @ sol_to_find
    ker, helps = x.kernel(return_solve_helper=True)
    try:
        sol = x.solve_with_helper(b, helps)
        print(f"Solution found. Is valid: {(x@sol-b).is_zero()}")
    except:
        print("Solution not found!")
```

## Testing epi-mono decomposition

```python
for i in range(100):
    X = lin.Hom.rand_dim(5, 4, 4)
    L, R = X.epi_mono()
    Lker = L.kernel()
    Xker = X.kernel()
    X_coker_dim = sum([(i+1)*d for i, d in enumerate(X.dim1)]) - sum([(i+1)*d for i, d in enumerate(Xker.dim1)])
    X_img_dim = sum([(i+1)*d for i, d in enumerate(L.dim1)])
    print(f"LR=X:{(L @ R - X).is_zero()}, L injective: {not any(Lker.dim1)}, Cokernel dim = Image dim: {X_coker_dim == X_img_dim}")
```
